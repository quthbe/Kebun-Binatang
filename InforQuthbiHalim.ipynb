```python
class Babi:
    def __init__(self, nama):
        self.nama = nama

    def ngook(self):
        print(f"{self.nama} berkata: Ngook... Ngook! (Nak Makan Woi)")


babi_baru = Babi("Piggyman")
babi_baru.ngook()
```
